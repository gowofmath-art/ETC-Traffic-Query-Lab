## M03A解壓縮

本筆記紀錄將 M03A 原始資料解壓縮並整理的過程。
也檢查過原始檔案數量正確，

設計基本流程如下：

1. 挑一天解壓縮 到 資料夾staging
2. 制定清洗規則，並寫好程式碼。 
3. 清洗一天，並將清洗完的資料放到Processed Data。
4. 以上步驟都沒問題的話，直接處理151天 (2026/1/1-2026/5/31)
